## Preparação do ambiente

Os caminhos são encontrados automaticamente quando o Jupyter é iniciado na raiz do projeto **ou** na pasta `notebooks`. Instale as bibliotecas com `python -m pip install -r requirements.txt` antes de executar.

```python
from pathlib import Path
import io
import sys
import pandas as pd

# Localiza a raiz do projeto a partir do diretório atual.
candidatos = [Path.cwd(), *Path.cwd().parents]
RAIZ = next((p for p in candidatos if (p / "data" / "dados_originais.csv").exists()), None)
if RAIZ is None:
    raise FileNotFoundError("Não encontrei data/dados_originais.csv. Abra o notebook dentro do projeto completo.")
if str(RAIZ) not in sys.path:
    sys.path.insert(0, str(RAIZ))
ARQUIVO_CSV = RAIZ / "data" / "dados_originais.csv"
print("Pasta do projeto:", RAIZ)
print("Versão do Pandas:", pd.__version__)
```

## Microatividade 1 — Ler um arquivo CSV

Importar o Pandas e carregar o arquivo utilizando o separador `;`, a engine `python` e codificação `utf-8`.

```python
dados_originais = pd.read_csv(
    ARQUIVO_CSV,
    sep=";",
    engine="python",
    encoding="utf-8",
    dtype={"Date": "string"},
)
print("Conjunto de dados lido com sucesso:")
print(dados_originais.to_string(index=False))
```

## Microatividade 2 — Criar um subconjunto de colunas

Selecionar três colunas da base importada, **sem alterar** o conjunto original.

```python
subconjunto = dados_originais[["ID", "Date", "Calories"]].copy()
print("Subconjunto com ID, Date e Calories:")
print(subconjunto.to_string(index=False))
```

## Microatividade 3 — Configurar máximo de linhas

Alterar a opção `display.max_rows` para **9999** e exibir o conteúdo completo com `to_string()`.

## Microatividade 5 — Informações gerais do DataFrame

Usar `info()`, dimensão, contagem de valores nulos, tipos de dados e memória consumida. A quantidade de bytes pode variar conforme a versão do Pandas e o ambiente.

## Trabalho Prático — Limpeza dos dados (passos 1 a 13)

**Passos 1 a 5:** importar o CSV obrigatório, conferir seu carregamento e visualizar a base. Esses passos já foram demonstrados nas microatividades 1, 4 e 5.

**Passo 6:** criar uma cópia independente para aplicar as mudanças somente nela.

```python
dados_trabalho = dados_originais.copy(deep=True)
print("Cópia criada. Registros:", len(dados_trabalho))
print("A base original não será alterada.")
```

### Passo 7 — Substituir valores nulos em Calories por zero

```python
print("Nulos em Calories antes:", dados_trabalho["Calories"].isna().sum())
dados_trabalho["Calories"] = pd.to_numeric(dados_trabalho["Calories"], errors="raise").fillna(0)
print("Nulos em Calories depois:", dados_trabalho["Calories"].isna().sum())
print(dados_trabalho.to_string(index=False))
```

### Passo 8 — Substituir temporariamente datas nulas por 1900/01/01

A instrução pede essa substituição intermediária. Em seguida tentaremos `to_datetime` com um formato explícito para identificar as inconsistências reais do conjunto.

```python
dados_trabalho["Date"] = dados_trabalho["Date"].fillna("1900/01/01")
print("ID 22 após preenchimento temporário:")
print(dados_trabalho.loc[dados_trabalho["ID"] == 22].to_string(index=False))

try:
    pd.to_datetime(dados_trabalho["Date"], format="%Y/%m/%d", errors="raise")
except (ValueError, TypeError) as erro:
    print("\nA primeira conversão apresenta erro devido a formatos de texto incompatíveis.")
    print("Detalhe:", str(erro).splitlines()[0])
```

### Passo 9 — Retirar o valor temporário e normalizar a grafia

**Observação técnica:** a data `1900/01/01` é *válida* para o formato `%Y/%m/%d`, ao contrário da afirmação do enunciado. Mesmo assim, ela precisa voltar a ser nula porque representa uma **data desconhecida**, não uma data real. Também removemos as aspas simples que constam literalmente na maioria dos valores do CSV.

O próximo teste expõe a data `20201226`, que não corresponde ao formato das demais.

```python
dados_trabalho["Date"] = dados_trabalho["Date"].str.strip().str.strip("'\"")
dados_trabalho["Date"] = dados_trabalho["Date"].replace({"1900/01/01": pd.NA})
print("ID 22 de volta a nulo:")
print(dados_trabalho.loc[dados_trabalho["ID"] == 22, ["ID", "Date"]].to_string(index=False))

try:
    pd.to_datetime(dados_trabalho["Date"], format="%Y/%m/%d", errors="raise")
except (ValueError, TypeError) as erro:
    print("\nSegunda tentativa: formato irregular identificado.")
    print("Detalhe:", str(erro).splitlines()[0])
```

### Passos 10 e 11 — Corrigir 20201226 e converter Date para datetime

Usamos `replace` para corrigir especificamente `20201226` e `pd.to_datetime` para transformar toda a coluna.

```python
dados_trabalho["Date"] = dados_trabalho["Date"].replace({"20201226": "2020/12/26"})
dados_trabalho["Date"] = pd.to_datetime(
    dados_trabalho["Date"], format="%Y/%m/%d", errors="raise"
)
print("Tipo de Date:", dados_trabalho["Date"].dtype)
print("ID 26 corrigido:")
print(dados_trabalho.loc[dados_trabalho["ID"] == 26].to_string(index=False))
print("\nBase com datas convertidas (ID 22 ainda aparece com NaT):")
print(dados_trabalho.to_string(index=False))
```

### Passos 12 e 13 — Remover o registro com Date ausente e validar a saída

`NaT` significa data ausente. Após remover os registros com `Date` nulo, o DataFrame terá **31 linhas**. Não há outras exclusões previstas no enunciado.

```python
dados_trabalho = dados_trabalho.dropna(subset=["Date"]).copy()
dados_trabalho["Calories"] = dados_trabalho["Calories"].astype("int64")
print("BASE FINAL TRATADA")
print(dados_trabalho.to_string(index=False))
print("\nRegistros:", len(dados_trabalho))
print("Nulos totais:", int(dados_trabalho.isna().sum().sum()))
print("IDs removidos:", sorted(set(dados_originais["ID"]) - set(dados_trabalho["ID"])))
assert len(dados_trabalho) == 31
assert dados_trabalho.isna().sum().sum() == 0
assert 22 not in dados_trabalho["ID"].tolist()
```

Gravar `resultados/dados_tratados.csv` e gerar o relatório textual; ambos podem ser usados como comprovação da execução. Também comparamos o resultado do notebook com a implementação principal em `src/tratamento.py`.
